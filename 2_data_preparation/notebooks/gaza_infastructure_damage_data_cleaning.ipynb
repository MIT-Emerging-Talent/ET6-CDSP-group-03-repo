### Load Data

```python
df = pd.read_json('data.json')
df.head()
```

#### First we need to convert the json format into csv

```python
data_frames = []
for idx, row in df.iterrows():
    civic_df = pd.json_normalize(row['civic_buildings'])
    educational_df = pd.json_normalize(row['educational_buildings'])
    places_of_worship_df = pd.json_normalize(row['places_of_worship'])
    residential_df = pd.json_normalize(row['residential'])

    data_frames.extend([civic_df, educational_df, places_of_worship_df, residential_df])

data = pd.concat(data_frames, ignore_index=True)
# data['report_date'] = df['report_date'].repeat(4).reset_index(drop=True)
data
```

### Handling Outliers

```python
Q1 = df["numeric_column"].quantile(0.25)
Q3 = df["numeric_column"].quantile(0.75)
IQR = Q3 - Q1
condition = ~(
    (df["numeric_column"] < (Q1 - 1.5 * IQR))
    | (df["numeric_column"] > (Q3 + 1.5 * IQR))
)
df = df[condition]
```
