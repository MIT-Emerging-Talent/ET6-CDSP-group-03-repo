
# 🧹 Dataset Cleaning Notebook

This notebook contains the steps used to clean **dataset1b**. The goal is to ensure the dataset is ready for analysis by addressing issues like missing values, duplicates, and inconsistent formatting.

We use Python's `pandas` library to handle the data cleaning tasks.



## 📥 Step 1: Load Dataset

In this step, we load the dataset into a pandas DataFrame so we can inspect and manipulate it.


```python
import pandas as pd  # noqa: F401
import json  # noqa: F401


# Load JSON file
with open(
    "../../1_datasets/data/01_category_war_events_data/gaza_displacement_orders_gazamaps/gaza-displacement.json"
) as f:
    data = json.load(f)

# Convert to DataFrame
df = pd.DataFrame(data)
df.head()
```


## 🧼 Step 2: Initial Cleaning

Here, we handle issues such as:
- Removing duplicates
- Dropping or imputing missing values
- Standardizing text data (e.g., column names or category labels)


```python
df.info()
df.isnull().sum()
```

```python
df["date"] = pd.to_datetime(df["date"])
df["area_sq_km_displacement"] = pd.to_numeric(
    df["area_sq_km_displacement"], errors="coerce"
)
df["area_sq_km_labeled_safe"] = pd.to_numeric(
    df["area_sq_km_labeled_safe"], errors="coerce"
)
```

```python
df.info()
df.isnull().sum()
```

```python
# Save to JSON file
df.to_json("gaza_displacement_cleaned.json", orient="records", indent=2)
```
