```python
import pandas as pd  # noqa: F401
import numpy as np  # noqa: F401
import matplotlib.pyplot as plt  # noqa: F401
import seaborn as sns  # noqa: F401
from pandas_profiling import ProfileReport  # noqa: F401
```

### Load Data

```python
df = pd.read_csv("dataset.csv")
df.head()
```

### Explore Data for missing values

```python
df.info()
df.describe()
df.isnull().sum()
```

### Handling Missing Values

```python
df["numeric_column"] = df["numeric_column"].fillna(df["numeric_column"].median())
df = df.dropna(subset=["critical_categorical_column"])
```

### Handling Duplicates

```python
df = df.drop_duplicates()
```

### Handling Outliers

```python
Q1 = df["numeric_column"].quantile(0.25)
Q3 = df["numeric_column"].quantile(0.75)
IQR = Q3 - Q1
condition = ~(
    (df["numeric_column"] < (Q1 - 1.5 * IQR))
    | (df["numeric_column"] > (Q3 + 1.5 * IQR))
)
df = df[condition]
```

### Convert Data Types

```python
df["date_column"] = pd.to_datetime(df["date_column"])
```

### Feature Engineering

```python
df["new_feature"] = df["feature1"] / df["feature2"]
```

### Data Validation

```python
df["category_column"].unique()
```

### Save Cleaned Data

```python
df.to_csv("cleaned_dataset.csv", index=False)
```
