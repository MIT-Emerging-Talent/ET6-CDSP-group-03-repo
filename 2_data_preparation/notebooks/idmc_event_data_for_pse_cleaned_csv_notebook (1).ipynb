  **idmc_event_data_for_pse_cleaning**

**1. Import Libraries**

```python
import pandas as pd  # noqa: F401
```

**2. Load the Dataset**

```python
df = pd.read_csv("../../1_datasets/data/01_category_war_events_data/gaza_diplacement_orders/idmc-event-data-for-pse.csv")
df.head()
```

**3.Data Cleaning**

```python
df = df.drop(index=0).reset_index(drop=True)
df.head()
```

**4. Data Overview**

```python
df.info()
```

```python
df.shape
```

```python
df = df.drop(
    columns=[
        "event_codes",
        "event_code_types",
        "type",
        "subtype",
        "category",
        "subcategory",
        "old_id",
    ],
)
```

```python
df.info()
```

```python
df["figure"] = pd.to_numeric(df["figure"], errors="coerce")
date_cols = [
    "displacement_date",
    "displacement_start_date",
    "displacement_end_date",
    "event_start_date",
    "event_end_date",
    "created_at",
]
df[date_cols] = df[date_cols].apply(pd.to_datetime, errors="coerce")
```

```python
df.info()
```

```python
df["latitude"] = pd.to_numeric(df["latitude"], errors="coerce")
df["longitude"] = pd.to_numeric(df["longitude"], errors="coerce")
df.info()
```

```python
# Save cleaned DataFrame to CSV (no index)
df.to_csv("cleaned_data.csv", index=False)

# The file "cleaned_data.csv" is now saved in your working directory.
# You can download it manually if needed.
```
