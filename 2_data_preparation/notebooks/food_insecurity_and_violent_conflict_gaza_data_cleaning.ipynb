```python
# doing necessary importation
import pandas as pd  # noqa: F401
```

```python
# load data
df = pd.read_csv(
    "../../1_datasets/data/01_category_war_events_data/food_insecurity_and_violent_conflict_gaza_dataset/fivc_gaza_octobe.csv"
)
df.head()
```

```python
df.shape
```

Removing completely empty rows 

```python
df.info()
```

As we see there are columns that completely empty, thats maybe due to excel empty columns

```python
# Calculate how many completely empty row are there:
# df[] to get it in a dataframe
# df.isna() creates a boolean DataFrame of the same shape, with True where values are NaN
# .all() checks if all values in the same row are TRUE (basically isna)
# shape: get dimensions (rows, cols)
df[df.isna().all(axis=1)].shape
```

```python
df = df.dropna(how="all")
df.shape
```

```python
# get the precentage of null values for each column
# isna() creates a dataframe with T/F value for each cell in the dataframe
# .mean() gets the mean of TRUE values for each column
df.isna().mean() * 100
```

dropping completely empty columns

```python
df = df.dropna(axis=1, how="all")
df.shape
```

checking null values 

```python
df.isna().sum()
```

Location and Admin 2 has Nan values
substituting all Nan values in Location and Admin 2 with the word "unknown"

```python
df = df.fillna("unknown")
df
# there is another way : df.fillna('value', inplace = True)
```

check that there is no more Nan values

```python
df.isnull().sum()
```

check  Admin 2 column 

```python
df["Admin 2"].value_counts()
```

we have:

* two Rafah
* Gaza city is the same as Gaza Governorate
* Deir el-Balah is the same as Deir al-Balah Governorate

so we are going to use replace method in pandas

```python
df.replace(
    {
        "Admin 2": {
            "Rafah ": "Rafah",
            "Gaza Governorate": "Gaza City",
            "Deir al-Balah Governorate": "Deir el-Balah",
        },
    },
    inplace=True,
)

df
```

check Admin 2 value count again

```python
df["Admin 2"].value_counts()
```

getting information about the data and the type of the values

```python
df.info()
# you can use df.dtypes()
```

after checking we can see that the date must be changed to type datetime instead of type object

```python
# Fix the date format manually
df["Date"] = df["Date"].replace("28-10-2023", "10/28/2023")


# converting  the column
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
# check
df.info()
```

checking if there is any empty cells in the Date column after converting its type 

```python
df[df["Date"].isna()]
```

good converted with no problems, now we will sort the messy dates so that we can see the events from the oldest to the newest

```python
df = df.sort_values(by="Date", ascending=True)
df
```

finally, we will rename some columns to make their names easier to understand

```python
df = df.rename(
    columns={
        "ID": "event_id",
        "Admin 1": "administrative_division_1",
        "Admin 2": "administrative_division_2",
        "Reported Means of Action": "Method of Attack",
    }
)
df.head()
```

```python
df.to_csv(
    "../../1_datasets/data/clean_datasets/fivc_gaza_october_cleaned.csv",
    index=False,
)
```
